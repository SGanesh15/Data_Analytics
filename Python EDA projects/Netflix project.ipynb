# Univariate Analysis

## Type Distribution

```python
sns.countplot(data=df, x='type', palette='Set2')
plt.title('Distribution of Content Type')
plt.show()
```

 ## 🔍Netflix has a higher number of TV Shows compared to Movies.
 

## Rating Distribution

```python
plt.figure(figsize=(14,7))
sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index, palette='Set3')
plt.xticks(rotation=45)
plt.title('Content Rating Distribution')
plt.show()
```

## 🔍Most Netflix content is rated TV-MA, indicating a lot of content for mature audiences.


## Release Year Distribution

```python
sns.histplot(df['release_year'], bins=30, kde=True, color='coral')
plt.title('Content Release Year Distribution')
plt.show
```

## Top 10 Countries

```python
top_countries=df['country'].value_counts().head(10)
top_countries.plot(kind='barh', color='skyblue')
plt.title('Top 10 Countries Producing Content')
plt.gca().invert_yaxis()
plt.show()
```

## Top Genres

```python
from collections import Counter
```

```python
genre_counter=Counter([genre.strip() for sublist in df['listed_in'].dropna().str.split(',') for genre in sublist])
top_genres=pd.DataFrame(genre_counter.most_common(10), columns=['Genre', 'Count'])

sns.barplot(data=top_genres, x='Count', y='Genre', palette='coolwarm')
plt.title('Top 10 Genres')
plt.show()
```

# Bivariate Analysis

## Type vs Release Year

```python
import matplotlib.ticker as ticker
```

```python
plt.figure(figsize=(16,8))
sns.countplot(data=df, x='release_year', hue='type', palette='viridis', dodge=True)

plt.title('Netflix Content Type by Release Year', fontsize=16)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Count', fontsize=12)

plt.xticks(rotation=45)

ax=plt.gca()
ax.xaxis.set_major_locator(ticker.MultipleLocator(5))

plt.legend(title='Type')
plt.tight_layout()
plt.show()
```

## 🔍The number of TV shows surged after 2015, aligning with Netflix’s strategy to produce original series.


## Type vs Rating

```python
sns.countplot(data=df, y='rating', hue='type', order=df['rating'].value_counts().index, palette='Set1')
plt.title('Type vs Content Rating')
plt.show()
```

## 🔍Movies on Netflix have more variation in ratings compared to TV shows.


# Correlation Analysis

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='YlGnBu')
plt.title('Correlation Heatmap')
plt.show()
```

## 🔍Release Year and Duration show very low correlation, as expected.

# Data cleaning

```python
df.isnull().sum()
```

```python
df['director']=df['director'].fillna('Unknown')
```

```python
df['cast']=df['cast'].fillna('Unknown')
```

```python
df['country']=df['country'].fillna('Unknown')
```

```python
df['date_added']=df['date_added'].fillna('Unknown')
```

```python
df['rating']=df['rating'].fillna('Unknown')
```

```python
df['duration']=df['duration'].fillna('Unknown')
```

```python
df[['duration_int', 'duration_type']]=df['duration'].str.extract(r'(\d+)\s*(\w+)')
```

```python
df['duration_int']=df['duration_int'].fillna(0).astype(int)
```

```python
df['duration_int']=pd.to_numeric(df['duration_int'], errors='coerce')
```

```python
df['duration_type']=df['duration_type'].fillna('unknown')
```

# Data cleaning

```python
df['price']=df['price'].replace('[\$,]', '', regex=True).astype(float)
```

```python
df['last_review']=pd.to_datetime(df['last_review'],errors='coerce')
```

```python
df['reviews_per_month']=df['reviews_per_month'].fillna(0)
```

```python
df.duplicated().sum()
```

# Univariate Analysis

## Price Distribution

```python
plt.figure(figsize=(12,6))
sns.histplot(df['price'],bins=100,kde=True)
plt.title('Price Distribution')
plt.xlim(0,1000)
plt.show()

print("\nPrice Statistics:")
print(df['price'].describe())
```

## 🔍 Insights:
- Most listings are priced between 50-200
- There are some extremely high-priced listings (up to $10,000)
- The average price is around 152withamedianof106


## Room Type Distribution

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df, x='room_type', order=df['room_type'].value_counts().index)
plt.title('Distribution of Room Types')
plt.xticks(rotation=45)
plt.show()

print("\nRoom Type Counts:")
print(df['room_type'].value_counts())
```

## 🔍 Insights:
- Entire home/apt is the most common listing type (52%)
- Private rooms make up 45% of listings
- Shared rooms are rare (2%)


## Neighbourhood Distribution

```python
plt.figure(figsize=(12,6))
sns.countplot(data=df, y='neighbourhood_group', order=df['neighbourhood_group'].value_counts().index)
plt.title('Listings by Borough')
plt.show()

print("\nNeighbourhood Group counts:")
print(df['neighbourhood_group'].value_counts())
```

## 🔍 Insights:
- Manhattan has the most listings (44%)
- Brooklyn follows closely (41%)
- Other boroughs have significantly fewer listings

# Bivariate Analysis

## Price by Neighbourhood Group

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='neighbourhood_group', y='price')
plt.ylim(0,500)
plt.title('Price Distribution by Borough')
plt.show()

print("\nAverage Price by Borough:")
print(df.groupby('neighbourhood_group')['price'].mean().sort_values(ascending=False))
```

## 🔍 Insights:
- Manhattan has the highest average prices ($196)
- Brooklyn follows with $124 average
- Bronx has the lowest average prices ($87)

## Price by Room Type

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df, x='room_type', y='price')
plt.ylim(0,500)
plt.title('Price Distribution by Room Type')
plt.show()

print("\nAverage Price by Room Type:")
print(df.groupby('room_type')['price'].mean().sort_values(ascending=False))
```

## 🔍 Insights:
- Entire homes/apartments are most expensive ($211 average)
- Private rooms average $89
- Shared rooms are cheapest ($70 average)

## Availability Analysis

```python
plt.figure(figsize=(12,6))
sns.histplot(df['availability_365'], bins=50, kde=True)
plt.title('Availability Throughout the Year')
plt.show()

print("\nAvailability Statistics:")
print(df['availability_365'].describe())
```

## 🔍 Insights:
- Many listings are available year-round (365 days)
- Median availability is 322 days
- Some listings have very limited availability (minimum 0 days)

# Temporal Analysis

## Last Review Date Analysis

```python
df['review_year']=df['last_review'].dt.year
df['review_month']=df['last_review'].dt.month

plt.figure(figsize=(12,6))
sns.countplot(data=df, x='review_year')
plt.title('Reviews Over Time')
plt.show()
```

## 🔍 Insights:
- Most reviews occurred in 2019
- Review activity increased significantly from 2011 to 2019
- Some listings haven't received reviews (missing last_review)

# Correlation Analysis

```python
numerical_cols = ['price', 'minimum_nights', 'number_of_reviews', 
                 'reviews_per_month', 'calculated_host_listings_count',
                 'availability_365']

plt.figure(figsize=(10,8))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

## 🔍 Insights:
- Number of reviews and reviews per month are strongly correlated (0.59)
- Price has weak negative correlation with number of reviews (-0.03)
- Availability shows weak positive correlation with price (0.10)

# Advanced Analysis

## Top Hosts by Listings Count

```python
top_hosts = df['host_id'].value_counts().head(10)
plt.figure(figsize=(12,6))
top_hosts.plot(kind='bar')
plt.title('Top 10 Hosts by Number of Listings')
plt.xlabel('Host ID')
plt.ylabel('Number of Listings')
plt.show()

print("\nTop Hosts:")
print(top_hosts)
```

## 🔍 Insights:
- Some hosts manage hundreds of listings
- The top host has 327 listings
- This suggests professional property management companies using Airbnb

## Price vs. Location (Geospatial Analysis)

```python
plt.figure(figsize=(12,8))
sns.scatterplot(data=df, x='longitude', y='latitude', hue='price',
                palette='viridis', alpha=0.6, size='price', sizes=(10,200))
plt.title('Price Distribution by Geographic Location')
plt.show()
```

## 🔍 Insights:
- Higher prices concentrated in Manhattan
- Brooklyn shows more moderate pricing
- Outliers with very high prices appear across all boroughs

# Outlier Detection

```python
high_price = df[df['price'] > 500]
print("\nHigh-Priced Listings (>$500):", len(high_price))

print("\nListings with Minimum Nights > 30:")
print(df[df['minimum_nights'] > 30]['minimum_nights'].value_counts().head(10))
```

## 🔍 Insights:
- 1,168 listings priced above $500
- Some listings require very long minimum stays (up to 1,250 nights)
- This suggests some listings may be targeting long-term rentals

# 🏁 Conclusion
The analysis reveals that Manhattan and Brooklyn dominate Airbnb listings in both volume and pricing, with entire homes/apartments being the most expensive and common type. High availability is observed for many listings, while some are either long-term rentals or priced excessively high. Correlation analysis shows limited relationships among variables, but highlights hosts with large portfolios—suggesting the presence of commercial operators. These insights can inform pricing strategies, neighborhood targeting, and policy considerations for Airbnb stakeholders.

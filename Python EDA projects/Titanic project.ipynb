# Handling Fare Outliers using IQR method

```python
Q1 = dt["Fare"].quantile(0.25)
Q3 = dt["Fare"].quantile(0.75)
IQR = Q3 - Q1
fare_upper_bound = Q3 + 1.5 * IQR
dt["Fare"] = dt["Fare"].apply(lambda x: fare_upper_bound if x > fare_upper_bound else x)    
```

```python
dt.Fare.describe()
```

```python
sns.boxplot(y=dt["Fare"])
plt.title("Fare Outliers")
plt.show()
```

### 🔍 Insights:
- Interquartile Range (IQR) applied → Used to cap extreme high values.
- Upper bound limit set at Q3 + 1.5 * IQR to handle outliers.
- Boxplot shows a reduced range → Outliers beyond this threshold were replaced.
- Fare values are now within a reasonable range → Prevents skewness in analysis.

# Visualizing survival rates by class and gender

```python
fig, ax = plt.subplots(figsize=(6,5))
sns.barplot(x="Pclass", y="Survived", data=dt, ax=ax, errorbar=None)
ax.set_title("Survival Rate by Passenger Class")
```

### 🔍 Insights:
- First-class passengers had the **highest survival rate**, while third-class passengers had the **lowest**.
- This indicates that higher-class passengers had better access to lifeboats.
- Socio-economic status played a significant role in survival chances.


```python
fig, ax = plt.subplots(figsize=(6,5))
sns.barplot(data=dt, x="Sex", y="Survived", ax=ax, errorbar=None)
ax.set_title("Survival Rate by Gender")
plt.show()
```

### 🔍 Insights:
- The survival rate of **females is significantly higher** than that of males.
- This confirms that women were prioritized for lifeboats.
- Males had a much lower survival rate, likely because they gave up their seats or were not prioritized.


# Survival distribution by Age

```python
plt.figure(figsize=(10,5))
sns.histplot(dt[dt["Survived"] == 1]["Age"], bins=30, kde=True, color="green", label="Survived")
plt.title("Survival Distribution by Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.legend()
plt.show()
```

### 🔍 Insights:
- The distribution shows that younger passengers (around 20-30 years old) had a higher survival count.
- There is a sharp peak around the age of 20, indicating that many survivors belonged to this age group.
- The survival count decreases as age increases, but there are survivors across all age ranges.
  

```python
plt.figure(figsize=(10,5))
sns.histplot(dt[dt["Survived"] == 0]["Age"], bins=30, kde=True, color="red", label="Did not survive")
plt.title("Survival Distribution by Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.legend()
plt.show()
```

### 🔍 Insights:
- The age distribution of non-survivors also peaks around the early 20s, similar to the survivors' plot.
- However, the count of non-survivors is significantly higher than that of survivors in most age groups.
- The KDE curve suggests that middle-aged and older passengers had a higher likelihood of non-survival, but younger adults (20s-30s) still form the largest portion of non-survivors.

# Survival by Embarkation

```python
plt.figure(figsize=(5,5))
sns.barplot(x=["Q","S"], y=[dt["Q"].mean(),dt["S"].mean()])
plt.title("Survival Rate by Embarkation Point")
plt.xlabel("Embarkation Point")
plt.ylabel("Survival rate")
plt.show()
```

### 🔍 Insights:
- Higher survival rate for Southampton (S) passengers → Likely due to more first-class travelers.
- Lower survival rate for Queenstown (Q) passengers → Possibly more third-class passengers with limited access to lifeboats.
- Survival influenced by class & demographics → First-class had better survival chances.
